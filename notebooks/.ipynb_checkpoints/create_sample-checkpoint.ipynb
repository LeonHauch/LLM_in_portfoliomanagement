## Sample creation

This notebook creates a random subsample of the ppo_training dataset for testing purposes. It comprises two columns of Adj Close and Volume_Ratio, respectively for 120 data entries.


```python
import os
import pandas as pd

FULL_PATH = "../data/preprocessed/data_ppo.parquet"   
SAMPLE_PATH = "../data/sample/data_ppo_sample.parquet"  

os.makedirs("../data/sample", exist_ok=True)
```

```python
df = pd.read_parquet(FULL_PATH)
print(df.shape)
df.head()
```

```python
adj_cols = [c for c in df.columns if c.startswith("Adj Close_")]
vol_cols = [c for c in df.columns if c.startswith("Volume_Ratio_")]

tickers = [c.replace("Adj Close_", "") for c in adj_cols][:2] 
keep_adj = [f"Adj Close_{t}" for t in tickers]
keep_vol  = [f"Volume_Ratio_{t}" for t in tickers]

cols = keep_adj + keep_vol
small = df[cols].copy()

small = small.iloc[:120].copy()

small.to_parquet(SAMPLE_PATH)
SAMPLE_PATH
```

```python
from environment.portfolio_env import PortfolioEnv
env = PortfolioEnv(data_path=SAMPLE_PATH, lookback_window=60)
obs, info = env.reset()
print("obs len:", len(obs), "| step:", info["step"])
obs, reward, terminated, truncated, info = env.step(env.action_space.sample())
reward, terminated
```
